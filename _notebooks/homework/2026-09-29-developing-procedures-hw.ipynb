## Exercise 1: Give procedures descriptive names

The lesson asks for clearer names than `do_step`, `mix`, or `heat`. The three helpers below say what they do, and `prepare_dessert` names the overall goal. Parameters make the steps reusable for another ingredient or oven setting. A helper returns one step; the outer procedure collects the steps; only the caller displays them.


````python
# CODE_RUNNER: 3.13 Exercise 1 - Descriptive reusable procedures
def measure_ingredient(ingredient, cups):
    return "Measure " + str(cups) + " cups of " + ingredient + "."

def mix_dessert_batter(ingredient):
    return "Mix " + ingredient + " with sugar, eggs, and milk."

def bake_dessert(temperature_f, minutes):
    return "Bake at " + str(temperature_f) + " F for " + str(minutes) + " minutes."

def prepare_dessert(ingredient, cups, temperature_f, minutes):
    return [measure_ingredient(ingredient, cups),
            mix_dessert_batter(ingredient),
            bake_dessert(temperature_f, minutes)]

for step in prepare_dessert("flour", 2, 350, 25):
    print(step)
print("Changed arguments:")
for step in prepare_dessert("oat flour", 3, 325, 30):
    print(step)
````

## Exercise 3: Split a long procedure and test the parts

An all-in-one dessert function might validate servings, scale every ingredient, build instructions, and print—all in one body. Splitting it makes each claim independently testable: valid servings, exact scaling, and readable instructions. The orchestration procedure combines the tested parts and returns a result instead of depending on printed text.

**Prediction:** A recipe using 2 cups flour and 1 cup milk for 4 servings needs 4 and 2 cups for 8 servings. The original recipe should remain unchanged. Zero or negative servings should fail explicitly rather than silently producing a bad recipe.


````python
# CODE_RUNNER: 3.13 Exercise 3 - Decompose and verify
def validate_servings(servings):
    if type(servings) is not int or servings <= 0:
        raise ValueError("Servings must be a positive whole number")

def scale_ingredients(recipe, base_servings, requested_servings):
    validate_servings(base_servings)
    validate_servings(requested_servings)
    multiplier = requested_servings / base_servings
    scaled = {}
    for ingredient, cups in recipe.items():
        scaled[ingredient] = cups * multiplier
    return scaled

def format_recipe(ingredients, servings):
    lines = ["Dessert servings: " + str(servings)]
    for ingredient, cups in ingredients.items():
        lines.append(ingredient + ": " + str(cups) + " cups")
    return lines

def prepare_scaled_recipe(recipe, base_servings, requested_servings):
    ingredients = scale_ingredients(recipe, base_servings, requested_servings)
    return format_recipe(ingredients, requested_servings)

recipe = {"flour": 2, "milk": 1}
for line in prepare_scaled_recipe(recipe, 4, 8):
    print(line)
assert scale_ingredients(recipe, 4, 8) == {"flour": 4.0, "milk": 2.0}
assert scale_ingredients(recipe, 4, 2) == {"flour": 1.0, "milk": 0.5}
assert recipe == {"flour": 2, "milk": 1}
assert format_recipe({"flour": 4.0}, 8) == ["Dessert servings: 8", "flour: 4.0 cups"]
for invalid in [0, -1, 2.5, True]:
    try:
        validate_servings(invalid)
    except ValueError:
        print("Rejected invalid servings:", invalid)
    else:
        raise AssertionError("Invalid serving count was accepted")
print("Scaling, formatting, unchanged input, and validation checks: PASS")
````

## Procedure abstraction and AP connection

Developing a procedure means choosing its responsibility, parameters, body, and return value. Calling one supplies arguments and uses its behavior. A caller can use `prepare_scaled_recipe` through its documented inputs without reproducing the scaling details. Descriptive names improve readability; small functions make tests more focused. Neither guarantees correctness or faster execution.

In AP notation, `PROCEDURE scaleIngredients(recipe, base, requested) { ... }` declares the operation; `RETURN(result)` ends that call and sends data back. The Python `return` statements serve the same role. An internal representation can change while the public input/output contract stays consistent.
