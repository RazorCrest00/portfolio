## Popcorn 1: Extract the hidden message

**Prediction:** The recovered words are `SIGNAL`, `CONFIRMED`, and `DELTA`. There are 31 characters, including the decoys.

| Piece | Start | Stop (excluded) | Expression |
| --- | --- | --- | --- |
| SIGNAL | 2 | 8 | `transmission[2:8]` |
| CONFIRMED | 11 | 20 | `transmission[11:20]` |
| DELTA | −5 | End | `transmission[-5:]` |

The negative index counts from the end, so the final word can be recovered without calculating its positive start index.

```python
# CODE_RUNNER: Popcorn Hack 1 - Extract the hidden message using indexing and slicing

transmission = "xxSIGNALxxxCONFIRMEDxx77xxDELTA"

word_1 = transmission[2:8]
print("Word 1:", word_1)

word_2 = transmission[11:20]
print("Word 2:", word_2)

word_3 = transmission[-5:]
print("Word 3:", word_3)

print("Total length:", len(transmission))
print("Every other character:", transmission[::2])
```

**Explanation:** `[2:8]` includes positions 2 through 7. The step-2 slice takes positions 0, 2, 4, and so on across the entire transmission; it does not restart at each word. It retains fragments such as `DLA` from `DELTA`, but none of the three full target words. These operations read the original string without modifying it.

## Popcorn 2: Clean and search

**Prediction:** The search returns `True` and the cleaned message contains five whitespace-separated words. The leading dots remain because the task only asks to remove `>`, `-`, and `!`.

```python
# CODE_RUNNER: Popcorn Hack 2 - Clean and search a noisy transmission using string methods

noisy = "   ...AGENT>>>Falcon---reports: the PACKAGE has ARRIVED!!!   "

trimmed = noisy.strip()
print("Trimmed:", repr(trimmed))

cleaned = trimmed.replace(">", "").replace("-", "").replace("!", "")
print("Cleaned:", cleaned)

lowered = cleaned.lower()
print("Lowered:", lowered)

has_arrived = "arrived" in lowered
print("Contains 'arrived':", has_arrived)

words = cleaned.split()
print("Word count:", len(words))
```

**Explanation:** `.strip()` removes outer whitespace; it does not remove internal spaces. Deleting separators joins `AGENT`, `Falcon`, and `reports:` into one token, so `.split()` counts five words rather than seven. A lowercase copy makes the search case-insensitive while preserving the cleaned text's original capitalization. `noisy` remains unchanged.

## Popcorn 3: Build a decoded report

**Prediction:** Both confidence lines say `Confidence: 87%`. The fragments become `ALPHA-SEVEN-DELTA`, and the final report combines the values into one readable line.

```python
# CODE_RUNNER: Popcorn Hack 3 - Build a decoded report using concatenation and formatting

agent_name = "Falcon"
target_word = "MIDNIGHT"
confidence = 87
fragments = ["ALPHA", "SEVEN", "DELTA"]

headline = "Agent " + agent_name + " intercepted: " + target_word
print(headline)

confidence_line = "Confidence: " + str(confidence) + "%"
print(confidence_line)

confidence_line_fstring = f"Confidence: {confidence}%"
print(confidence_line_fstring)

joined_fragments = "-".join(fragments)
print("Joined fragments:", joined_fragments)

final_report = f"REPORT: {headline} | {confidence_line} | Fragments: {joined_fragments}"
print(final_report)
```

**Explanation:** Concatenation requires string operands, so `str(confidence)` is necessary when using `+`. An f-string formats the integer directly. `.join()` puts the separator between fragments without a trailing dash. The strings and list used as inputs are not changed.

## Homework: Crack the Final Transmission

**Prediction:** Splitting produces four fields. The sliced year is `2026`, the year check is `True`, and the case-insensitive search finds `dawn`. The decoded message is `THE EAGLE LANDS AT DAWN`.

| Field | Index | Value |
| --- | --- | --- |
| Date | 0 | 2026-09-15 |
| Agent | 1 | GHOST |
| Message code | 2 | THE-EAGLE-LANDS-AT-DAWN |
| Confidence | 3 | 87 |

The confidence begins as text because `.split()` returns strings. It is displayed with `%` in the Markdown row, but emitted as an unquoted number in JSON.

```python
# CODE_RUNNER: Homework - Crack the Final Transmission using slicing, methods, and formatting

raw_transmission = "  2026-09-15,GHOST,THE-EAGLE-LANDS-AT-DAWN,87  "

# 1. Remove outer whitespace and split the four comma-separated fields.
segments = raw_transmission.strip().split(",")
print("Segments:", segments)

# 2. Extract fields by their zero-based positions.
date = segments[0]
agent = segments[1]
message_code = segments[2]
confidence_str = segments[3]
print("Date:", date, "| Agent:", agent, "| Message code:", message_code, "| Confidence:", confidence_str)

# 3. Slice the year before comparing; the stop index is excluded.
year = date[:4]
year_check = year == "2026"
print("Year:", year)
print("Year check:", year_check)

# 4-5. Build a new message and search a lowercase copy.
message = message_code.replace("-", " ")
found_dawn = "dawn" in message.lower()
print("Decoded message:", message)
print("Found 'dawn':", found_dawn)

# 6. Format the decoded fields as a Markdown table row.
markdown_row = f"| {date} | {agent} | {message} | {confidence_str}% |"
print("Markdown row:", markdown_row)

# 7. Use f-strings for the fields and literal strings for the outer braces.
json_line = (
    "{"
    f'"date": "{date}", "agent": "{agent}", "message": "{message}", '
    f'"confidence": {confidence_str}, "contains_dawn": {str(found_dawn).lower()}'
    "}"
)
print("JSON line:", json_line)
```

### Why the result is correct

- `date[:4]` extracts exactly the year; comparing it with `"2026"` yields a Boolean. Using `date[4:]` would extract the remainder instead.
- `.replace("-", " ")` inserts spaces rather than deleting word boundaries. `.lower()` changes only the search copy, so the report preserves the message's capitalization.
- `in` checks for a **substring**, as the assignment requests. It also matches `predawn`; it is not a whole-word test.
- JSON keys and text values use double quotes. The confidence is numeric, and `str(found_dawn).lower()` produces unquoted `true` or `false`. Python's `True` would not be valid JSON.
- Each transformation creates a new string; the raw transmission remains available for comparison.

**Scope:** This solution handles the assignment's simple four-field CSV-style input. Plain `.split(",")` is not a general CSV parser: quoted commas need a proper parser. The hand-built JSON assumes these controlled text fields contain no quotes, backslashes, or control characters. General serialization would normally use a JSON library; this assignment explicitly prohibits imports.

## Additional verification: changed inputs

The following independent runner extracts the same operations into a small function so eight different inputs can be checked without editing the required solution. These are synthetic test cases, not claims about real intercepted messages.

| Case | Purpose |
| --- | --- |
| Required sample | Confirm the original data and format |
| Outer tabs/newlines | Verify `.strip()` handles whitespace beyond spaces |
| Mixed case | Verify case-insensitive search without changing displayed case |
| No dawn | Exercise the false branch and JSON `false` |
| Different year | Verify the slice comparison is not always true |
| Substring | Confirm `predawn` follows the required `in` semantics |
| Zero confidence | Preserve a valid zero value |
| Full confidence | Preserve 100 rather than assuming two digits |

Every case checks the fields, year result, search result, Markdown row, JSON line, and preservation of the original input. A failed assertion stops execution instead of printing a misleading success message.

```python
# CODE_RUNNER: Extension - Verify eight changed inputs and string immutability

def decode_transmission(raw):
    segments = raw.strip().split(",")
    date = segments[0]
    agent = segments[1]
    message_code = segments[2]
    confidence_str = segments[3]
    year_check = date[:4] == "2026"
    message = message_code.replace("-", " ")
    found_dawn = "dawn" in message.lower()
    markdown_row = f"| {date} | {agent} | {message} | {confidence_str}% |"
    json_line = (
        "{"
        f'"date": "{date}", "agent": "{agent}", "message": "{message}", '
        f'"confidence": {confidence_str}, "contains_dawn": {str(found_dawn).lower()}'
        "}"
    )
    return segments, year_check, message, found_dawn, markdown_row, json_line

# The expected values below are test oracles, not hardcoded program results.
cases = [
    ("required sample", "  2026-09-15,GHOST,THE-EAGLE-LANDS-AT-DAWN,87  ",
     ["2026-09-15", "GHOST", "THE-EAGLE-LANDS-AT-DAWN", "87"], True, "THE EAGLE LANDS AT DAWN", True),
    ("outer whitespace", "\t2026-09-15,GHOST,DAWN,87\n",
     ["2026-09-15", "GHOST", "DAWN", "87"], True, "DAWN", True),
    ("mixed case", "2026-10-01,HAWK,Meet-At-DaWn,72",
     ["2026-10-01", "HAWK", "Meet-At-DaWn", "72"], True, "Meet At DaWn", True),
    ("no dawn", "2026-10-01,HAWK,MEET-AT-NOON,72",
     ["2026-10-01", "HAWK", "MEET-AT-NOON", "72"], True, "MEET AT NOON", False),
    ("different year", "2027-01-01,OWL,ARRIVE-AT-DAWN,50",
     ["2027-01-01", "OWL", "ARRIVE-AT-DAWN", "50"], False, "ARRIVE AT DAWN", True),
    ("substring", "2026-10-01,OWL,PREDAWN-CHECK,50",
     ["2026-10-01", "OWL", "PREDAWN-CHECK", "50"], True, "PREDAWN CHECK", True),
    ("zero confidence", "2026-10-01,OWL,NO-SIGNAL,0",
     ["2026-10-01", "OWL", "NO-SIGNAL", "0"], True, "NO SIGNAL", False),
    ("full confidence", "2026-10-01,OWL,DAWN,100",
     ["2026-10-01", "OWL", "DAWN", "100"], True, "DAWN", True),
]

for label, raw, expected_fields, expected_year, expected_message, expected_dawn in cases:
    original = raw
    fields, year_ok, message, dawn_found, row, json_text = decode_transmission(raw)
    assert fields == expected_fields, label + ": fields"
    assert year_ok is expected_year, label + ": year"
    assert message == expected_message, label + ": message"
    assert dawn_found is expected_dawn, label + ": search"
    expected_row = "| " + " | ".join([
        expected_fields[0], expected_fields[1], expected_message, expected_fields[3] + "%"
    ]) + " |"
    assert row == expected_row, label + ": Markdown row"
    expected_json = (
        '{"date": "' + expected_fields[0] + '", "agent": "' + expected_fields[1]
        + '", "message": "' + expected_message + '", "confidence": ' + expected_fields[3]
        + ', "contains_dawn": ' + ("true" if expected_dawn else "false") + '}'
    )
    assert json_text == expected_json, label + ": JSON line"
    assert raw == original, label + ": original input"
    print("PASS:", label)

print("All", len(cases), "cases passed; 7 checks per case.")
```
