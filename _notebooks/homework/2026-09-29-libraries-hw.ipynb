## Popcorn 1: Read the API and change the values

**Changes:** Jersey range is **1–55**, the square area is **225** rather than 144, and the points list is `[18, 26, 22, 30, 24]`. **Prediction:** A jersey is an inclusive integer from 1 through 55; the court side is 15.0 feet; the average and rounded-up average are both 24. A random review score must belong to the list. A fixed seed makes this saved demonstration reproducible; it is not a claim that random outputs always repeat.


````python
# CODE_RUNNER: 3.14 Popcorn 1 - Changed APIs and inputs
import math
import random

print("randint API:", random.randint.__doc__.strip().splitlines()[0])
print("ceil API:", math.ceil.__doc__.strip().splitlines()[0])
random.seed(314)
points_per_game = [18, 26, 22, 30, 24]
print("Random game to review:", random.choice(points_per_game), "points")
jersey = random.randint(1, 55)
print("Random jersey number:", jersey)
side = math.sqrt(225)
print("Square court side:", side, "feet")
average = sum(points_per_game) / len(points_per_game)
print("Points per game:", average, "-> rounded up:", math.ceil(average))
assert 1 <= jersey <= 55
assert side == 15.0
````

## Popcorn 2: Three real Flask routes

**Changes:** The practice roster contains Adhvay, Ishan, and Rohan; the added `/record` route returns the fictional team's wins and losses. All routes are defined before creating the test client. `jsonify` constructs JSON responses, and `get_json` decodes their data. The test client exercises Flask routing without opening a network port.


````python
# CODE_RUNNER: 3.14 Popcorn 2 - Flask routes
from flask import Flask, jsonify
import json

app = Flask("uesl_homework", root_path="/")

@app.route("/score")
def score():
    return jsonify({"home": 6, "away": 4, "period": 2})

@app.route("/roster")
def roster():
    return jsonify({"players": ["Adhvay", "Ishan", "Rohan"]})

@app.route("/record")
def record():
    return jsonify({"team": "UESL Practice Team", "wins": 3, "losses": 1})

client = app.test_client()
for path in ["/score", "/roster", "/record"]:
    response = client.get(path)
    assert response.status_code == 200
    print(path + ":", json.dumps(response.get_json(), sort_keys=True))
assert client.get("/missing").status_code == 404
print("Unknown route correctly returns 404")
````

## Homework: UESL Rocket League season report

This original **simulated** season uses goals as the points statistic. The API returns totals, mean, best game, and combined points against each opponent. `statistics` calculates the mean, `json` serializes the report, and `datetime` calculates the countdown. No real player's performance is asserted.

**Prediction:** 30 total goals, 6.0 average, 8 best game; Comets 14, Falcons 9, Meteors 7. The next practice match is October 10, 2026: **11 days after the report date, September 29**. The explicit report date keeps saved and browser results reproducible; change `as_of` to `date.today()` for a live countdown.


````python
# CODE_RUNNER: 3.14 Homework - UESL season statistics
import json
import statistics
from datetime import date

def summarize_season(season):
    """Summarize games without modifying the input.

    Parameters:
        season: list of dictionaries with date (ISO string), opponent (string),
                and points (nonnegative number; goals in this simulated season).
    Returns:
        dict containing total_points, average_points rounded to one decimal,
        best_game (largest points value), and points_by_opponent totals.
        An empty season returns total 0, mean None, best None, and an empty map.
    """
    points = [game["points"] for game in season]
    by_opponent = {}
    for game in season:
        opponent = game["opponent"]
        by_opponent[opponent] = by_opponent.get(opponent, 0) + game["points"]
    return {
        "total_points": sum(points),
        "average_points": round(statistics.mean(points), 1) if points else None,
        "best_game": max(points) if points else None,
        "points_by_opponent": by_opponent,
    }

games = [
    {"date": "2026-09-10", "opponent": "Comets", "points": 6},
    {"date": "2026-09-14", "opponent": "Falcons", "points": 4},
    {"date": "2026-09-18", "opponent": "Comets", "points": 8},
    {"date": "2026-09-22", "opponent": "Meteors", "points": 7},
    {"date": "2026-09-26", "opponent": "Falcons", "points": 5},
]
as_of = date(2026, 9, 29)
next_game = date(2026, 10, 10)
report = summarize_season(games)
report["sport"] = "Rocket League (simulated goals)"
report["report_date"] = as_of.isoformat()
report["next_game"] = next_game.isoformat()
report["days_until_next_game"] = (next_game - as_of).days
print(json.dumps(report, indent=2, sort_keys=True))
requirements_txt = """# Hypothetical full app: table processing and Flask scoreboard.
# json, datetime, and statistics come with Python and are not pip packages.
pandas
Flask==3.1.3
"""
print("requirements.txt:")
print(requirements_txt)
````

## Tests: Empty seasons, opponent totals, and dates

Empty data has no meaningful average or best game, so those values are `None` (JSON `null`) rather than an invented zero-point game. A single game, repeated opponent, unchanged input, JSON round-trip, and the fixed date difference are checked below. The independent verifier also checks each Flask response body and its HTTP status.


````python
# CODE_RUNNER: 3.14 Tests - Season API and dates
import json
import statistics
from datetime import date

def summarize_season(season):
    """Summarize games without modifying the input.

    Parameters:
        season: list of dictionaries with date (ISO string), opponent (string),
                and points (nonnegative number; goals in this simulated season).
    Returns:
        dict containing total_points, average_points rounded to one decimal,
        best_game (largest points value), and points_by_opponent totals.
        An empty season returns total 0, mean None, best None, and an empty map.
    """
    points = [game["points"] for game in season]
    by_opponent = {}
    for game in season:
        opponent = game["opponent"]
        by_opponent[opponent] = by_opponent.get(opponent, 0) + game["points"]
    return {
        "total_points": sum(points),
        "average_points": round(statistics.mean(points), 1) if points else None,
        "best_game": max(points) if points else None,
        "points_by_opponent": by_opponent,
    }

empty = summarize_season([])
assert empty == {"total_points": 0, "average_points": None, "best_game": None, "points_by_opponent": {}}
print("Empty season:", json.dumps(empty, sort_keys=True))
games = [{"date": "2026-09-10", "opponent": "Comets", "points": 6},
         {"date": "2026-09-18", "opponent": "Comets", "points": 8}]
before = [game.copy() for game in games]
result = summarize_season(games)
assert result == {"total_points": 14, "average_points": 7.0, "best_game": 8, "points_by_opponent": {"Comets": 14}}
assert games == before
assert json.loads(json.dumps(result)) == result
assert summarize_season(games[:1])["average_points"] == 6.0
assert (date(2026, 10, 10) - date(2026, 9, 29)).days == 11
print("Opponent grouping, single game, unchanged input, JSON, countdown: PASS")
````
