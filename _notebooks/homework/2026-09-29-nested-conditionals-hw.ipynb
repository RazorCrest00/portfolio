## Popcorn 1: Trace all after-school paths

**Prediction:** With weekday `True` and homework finished `False`, the outer weekday branch runs and its inner `else` prints **Finish homework**. Weekday/finished prints **Free time**; a non-weekday prints **Weekend plan** and never evaluates the inner decision.


````python
# CODE_RUNNER: 3.07 Popcorn 1 - Three after-school paths
def after_school_plan(is_weekday, homework_finished):
    if is_weekday:
        if homework_finished:
            return "Free time"
        else:
            return "Finish homework"
    else:
        return "Weekend plan"

for weekday, finished in [(True, False), (True, True), (False, False)]:
    print("weekday =", weekday, "finished =", finished, "->", after_school_plan(weekday, finished))
````

## Popcorn 2: Repair the account guard

The starter has two sibling `if` statements, so a logged-out student can reach the permission check. Indenting that check under `if logged_in` makes account access the prerequisite. The outer `else` handles logged-out users; the inner `else` handles logged-in users who lack permission. The retained **Account found** line is additional context for either logged-in outcome.


````python
# CODE_RUNNER: 3.07 Popcorn 2 - Permission depends on login
def account_access(logged_in, has_permission):
    if logged_in:
        print("Account found")
        if has_permission:
            print("Page opened")
        else:
            print("Permission needed")
    else:
        print("Log in first")

for logged_in, has_permission in [(True, True), (True, False), (False, True), (False, False)]:
    print("login =", logged_in, "permission =", has_permission)
    account_access(logged_in, has_permission)
````

## Popcorn 3: Lunch first, then practice

**Prediction:** Both lunch branches contain a practice decision, creating four distinct recommendations. Lunch is checked first; practice is checked only within the chosen lunch branch. A compound `and` could test one combination, but it would not show all four paths by itself.


````python
# CODE_RUNNER: 3.07 Popcorn 3 - All four lunch paths
def lunch_plan(lunch_packed, practice_after_school):
    if lunch_packed:
        if practice_after_school:
            return "Bring packed lunch and an extra practice snack."
        else:
            return "Bring packed lunch; no practice snack needed."
    else:
        if practice_after_school:
            return "Buy lunch and an extra practice snack."
        else:
            return "Buy lunch; no practice snack needed."

for packed in [True, False]:
    for practice in [True, False]:
        print("packed =", packed, "practice =", practice, "->", lunch_plan(packed, practice))
````

## Homework: Student daily routine dashboard

The outer account decision gates all schedule advice. Once logged in, a homework branch contains a practice check, and a meal branch contains another practice check. A logged-out student receives only the login recommendation, so inner schedule information is not displayed before account access.

**Prediction for the starter:** Account ready; finish homework before practice; bring a packed lunch and a practice snack. This is a simulated dashboard decision, not a replacement for real authentication.


````python
# CODE_RUNNER: 3.07 Homework - Student daily routine
def daily_routine(student):
    recommendations = []
    if student["logged_in"]:
        recommendations.append("Account ready: review today's schedule.")
        if student["homework_finished"]:
            if student["practice_today"]:
                recommendations.append("Homework finished: attend practice, then dinner.")
            else:
                recommendations.append("Homework finished: enjoy free time before dinner.")
        else:
            if student["practice_today"]:
                recommendations.append("Finish homework before practice; leave time for dinner.")
            else:
                recommendations.append("Finish homework before dinner and free time.")
        if student["lunch_packed"]:
            if student["practice_today"]:
                recommendations.append("Bring packed lunch and an extra practice snack.")
            else:
                recommendations.append("Bring packed lunch; plan dinner after homework.")
        else:
            if student["practice_today"]:
                recommendations.append("Buy lunch and an extra practice snack.")
            else:
                recommendations.append("Buy lunch; plan dinner after homework.")
    else:
        recommendations.append("Log in first to view the student schedule.")
    return recommendations

student = {"logged_in": True, "homework_finished": False,
           "lunch_packed": True, "practice_today": True}
for recommendation in daily_routine(student):
    print(recommendation)
````

## Tests: Every Boolean combination

There are `2 × 2 × 2 × 2 = 16` input combinations. All eight logged-out cases must collapse to the login-only message; the eight logged-in cases exercise every homework/practice and lunch/practice branch. The independent checker also compares each message with an explicit expected result.


````python
# CODE_RUNNER: 3.07 Tests - All 16 student states
def daily_routine(student):
    recommendations = []
    if student["logged_in"]:
        recommendations.append("Account ready: review today's schedule.")
        if student["homework_finished"]:
            if student["practice_today"]:
                recommendations.append("Homework finished: attend practice, then dinner.")
            else:
                recommendations.append("Homework finished: enjoy free time before dinner.")
        else:
            if student["practice_today"]:
                recommendations.append("Finish homework before practice; leave time for dinner.")
            else:
                recommendations.append("Finish homework before dinner and free time.")
        if student["lunch_packed"]:
            if student["practice_today"]:
                recommendations.append("Bring packed lunch and an extra practice snack.")
            else:
                recommendations.append("Bring packed lunch; plan dinner after homework.")
        else:
            if student["practice_today"]:
                recommendations.append("Buy lunch and an extra practice snack.")
            else:
                recommendations.append("Buy lunch; plan dinner after homework.")
    else:
        recommendations.append("Log in first to view the student schedule.")
    return recommendations

case_count = 0
messages = set()
for logged_in in [False, True]:
    for finished in [False, True]:
        for packed in [False, True]:
            for practice in [False, True]:
                student = {"logged_in": logged_in, "homework_finished": finished,
                           "lunch_packed": packed, "practice_today": practice}
                result = daily_routine(student)
                if not logged_in:
                    assert result == ["Log in first to view the student schedule."]
                else:
                    assert len(result) == 3
                case_count += 1
                messages.update(result)
                print(str((logged_in, finished, packed, practice)) + ": " + " | ".join(result))
print("Cases checked:", case_count)
print("Distinct recommendations:", len(messages))
````
