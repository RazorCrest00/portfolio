The fixed cell uses `foods[3]` for the fourth print. Python positions are 0–3; the equivalent AP positions are 1–4. Therefore AP's invalid fifth position translates to Python index 4, and the corrected last position translates to index 3.

```python
# CODE_RUNNER: Lists Popcorn 1 - Correct the index
foods = ["pizza", "sushi", "tacos", "ramen"]

print(foods[0])   # Prediction: pizza
print(foods[-1])  # Prediction: ramen
print(len(foods)) # Prediction: 4
print(foods[3])   # Prediction: ramen; foods[4] would raise IndexError
```

**Discussion:** For a nonempty list of length `n`, counting from zero gives positions `0` through `n - 1`. `[-1]` selects the last element even if the list grows. An empty list has no last element, so the homework checks before indexing.

## Popcorn 2: Accumulate and average

**Prediction:** The sample values total `42`; dividing by six gives `7.0`. The running total progresses through `7 → 13 → 21 → 26 → 33 → 42`. Initialization belongs before the loop so earlier values are retained.

```python
# CODE_RUNNER: Lists Popcorn 2 - Total and safe average
numbers = [7, 6, 8, 5, 7, 9]  # Lesson sample, not collected class data.

total = 0
for number in numbers:
    total += number

print("Total:", total)
if len(numbers) == 0:
    print("No data yet")
else:
    average = round(total / len(numbers), 2)
    print("Average:", average)
```

```python
# CODE_RUNNER: Lists Popcorn 2 - Empty input check
numbers = []

total = 0
for number in numbers:
    total += number

print("Total:", total)
if len(numbers) == 0:
    print("No data yet")
else:
    average = round(total / len(numbers), 2)
    print("Average:", average)
```

## Popcorn 3: Count and filter the survey

**Prediction:** winter has 1 vote, summer 3, fall 2, and spring 1. The qualifying categories are `summer` and `fall`. This uses the lesson's sample responses and retains their first-seen category order.

```python
# CODE_RUNNER: Lists Popcorn 3 - Count then filter
seasons = ["winter", "summer", "fall", "summer", "spring", "summer", "fall"]

frequency = {}
for season in seasons:
    if season in frequency:
        frequency[season] += 1
    else:
        frequency[season] = 1

print(frequency)

popular = []
for season in frequency:
    if frequency[season] >= 2:
        popular.append(season)

print(popular)
```

**Discussion:** The dictionary handles a new category by creating a key during the loop. Separate counters would require a variable and matching logic for each anticipated category. The filter loops over distinct keys, so each popular category appears once. It appends to a new list and never removes or replaces an item in `seasons`.

The counts add to seven, matching the number of responses. A category receiving exactly two votes qualifies because the comparison is `>= 2`, not `> 2`.

## Homework: Snack votes and prices

**Prediction:** 12 responses, last response `cookies`, total `31`, counts `chips: 5`, `cookies: 4`, `fruit: 2`, `pretzels: 1`, and popular snacks `chips` and `cookies`.

**Average-format clarification:** The live starter and expected output request `int()` and show `Average: 2`; its rubric also asks for `round(..., 2)`. The solution keeps the unrounded average and calculates `rounded_average` as `2.58` for the rubric, then displays the integer version to match the specified six-line report. `int()` truncates; it does not round to the nearest integer.

The two lists describe the same responses. When changing the data, keep one price for each vote.

```python
# CODE_RUNNER: Lists Homework - Index, accumulate, count, filter
votes = ["chips", "cookies", "chips", "fruit", "cookies", "chips",
         "pretzels", "cookies", "chips", "fruit", "chips", "cookies"]
prices = [2, 3, 2, 4, 3, 2, 1, 3, 2, 4, 2, 3]

print(len(votes))
if len(votes) == 0:
    print("No votes yet")
else:
    print(votes[-1])

total = 0
for price in prices:
    total += price
print("Total:", int(total))

if len(prices) == 0:
    average = None
    rounded_average = None
    print("Average: No data yet")
else:
    average = total / len(prices)
    rounded_average = round(average, 2)
    print("Average:", int(average))

frequency = {}
for snack in votes:
    if snack in frequency:
        frequency[snack] += 1
    else:
        frequency[snack] = 1
print(frequency)

popular = []
for snack in frequency:
    if frequency[snack] >= 3:
        popular.append(snack)
print(popular)
```

### Trace and explanation

| Quantity | Evidence |
| --- | --- |
| Vote count | `len(votes)` is 12; it is calculated, not printed as a constant. |
| Last vote | `votes[-1]` remains correct when the list length changes. |
| Running total | `2 → 5 → 7 → 11 → 14 → 16 → 17 → 20 → 22 → 26 → 28 → 31` |
| Unrounded mean | `31 / 12 = 2.583333…` |
| Rounded mean | `round(31 / 12, 2) = 2.58` |
| Integer display | `int(31 / 12) = 2` |
| Frequency invariant | `5 + 4 + 2 + 1 = 12`; every response is counted once. |
| Filter boundary | At least 3 votes: chips and cookies qualify; fruit's 2 do not. |

`frequency` starts empty; its `if/else` distinguishes an existing category from a new one. `popular` starts as a separate empty list. The algorithms read `votes` and `prices` without modifying them.

**Why lists simplify this problem:** Adding another response changes the data, while the same loops still calculate the total, count categories, and apply the threshold. Separate variables and hand-written additions would require repeated code edits. This is a maintainability benefit, not a claim that every list-based program is faster.
