## Popcorn 1: Status reports with parameters

`report_status(location, status)` uses both inputs in one reusable print statement. The two calls change **both** arguments.

**Prediction:** The first line identifies Practice Zone A and its ready team. The second identifies Practice Zone B and its completed equipment check. Defining the function alone produces no message; its body executes when called.

````python
# CODE_RUNNER: Calling Procedures Popcorn 1 - Status reports

def report_status(location: str, status: str) -> None:
    print("Status update from " + location + ": " + status)

report_status("Practice Zone A", "Drill team ready")
report_status("Practice Zone B", "Equipment check complete")
````

### Parameters, arguments, and order

| Part | In this solution |
| --- | --- |
| Procedure name | `report_status` |
| Parameters | `location` and `status` in the definition |
| First call's arguments | `"Practice Zone A"` and `"Drill team ready"` |
| Binding | The first argument becomes `location`; the second becomes `status`. |
| Side effect | One status line is printed per call. |
| Return value | `None`, because this procedure does not explicitly return a value. |

The `str` and `None` annotations document expected inputs and the return type; they do not perform runtime validation. Reversing positional arguments changes their meaning. Keyword arguments such as `report_status(status="Ready", location="Practice Zone C")` can name the intended bindings explicitly.

## Popcorn 2: A helper inside an outer procedure

`is_priority(incident_type)` returns a Boolean. `process_report(location, incident_type)` calls that helper, stores the result, and selects one message.

**Prediction:** `"Medical Emergency"` produces the priority message; `"Power Outage"` produces the standard report. The helper itself prints nothing.

````python
# CODE_RUNNER: Calling Procedures Popcorn 2 - Helper and outer procedure

def is_priority(incident_type: str) -> bool:
    return incident_type == "Medical Emergency"


def process_report(location: str, incident_type: str) -> None:
    priority = is_priority(incident_type)
    if priority:
        print("PRIORITY: " + incident_type + " reported in " + location)
    else:
        print("STANDARD REPORT: " + incident_type + " reported in " + location)

process_report("Practice Zone A", "Medical Emergency")
process_report("Practice Zone B", "Power Outage")
````

### Why return is different from print

`return` sends a value back to the caller and ends that function call. Here, the comparison produces actual Booleans `True` or `False`, so the outer procedure can make a decision. Printing a Boolean instead would leave an implicit return value of `None`. Returning the string `"False"` would also be wrong: a nonempty string is truthy in Python.

The outer procedure owns the message, and the helper owns the classification rule. A future rule change belongs in the helper. The main code calls the outer procedure rather than copying its decision logic for each report.

## Homework: Dispatch-response simulation

The required names are `check_priority(incident_type)` and `dispatch_response(location, incident_type)`. The helper uses the lesson's exact `"Medical Emergency"` match. The outer procedure stores that Boolean and prints a different response for each branch.

Four calls exceed the minimum of three. The first three use three different incident types; the fourth repeats the priority type at a new location to demonstrate reuse without leftover state.

| Call | Arguments: location, incident type | Predicted branch |
| --- | --- | --- |
| 1 | Practice Zone A, Medical Emergency | Priority |
| 2 | Practice Zone B, Power Outage | Standard |
| 3 | Practice Zone C, Road Blockage | Standard |
| 4 | Practice Zone D, Medical Emergency | Priority |

````python
# CODE_RUNNER: Calling Procedures Homework - Dispatch responses

def check_priority(incident_type: str) -> bool:
    return incident_type == "Medical Emergency"


def dispatch_response(location: str, incident_type: str) -> None:
    priority = check_priority(incident_type)
    if priority:
        print("PRIORITY RESPONSE: " + incident_type + " reported in " + location)
    else:
        print("STANDARD RESPONSE: " + incident_type + " reported in " + location)

dispatch_response("Practice Zone A", "Medical Emergency")
dispatch_response("Practice Zone B", "Power Outage")
dispatch_response("Practice Zone C", "Road Blockage")
dispatch_response("Practice Zone D", "Medical Emergency")
````

### Trace the call and return

| Step | First call: Medical Emergency | Second call: Power Outage |
| --- | --- | --- |
| Bind outer parameters | `location` = Practice Zone A; `incident_type` = Medical Emergency | `location` = Practice Zone B; `incident_type` = Power Outage |
| Invoke helper | `check_priority("Medical Emergency")` | `check_priority("Power Outage")` |
| Helper evaluates equality | `True` | `False` |
| Return to caller | Store `True` in the local `priority` variable | Store `False` in the local `priority` variable |
| Select output | Print the priority response | Print the standard response |
| Finish outer call | Return `None`; continue with the next call | Return `None`; continue with the next call |

Each call binds its own arguments. No global location, incident variable, or stored priority flag is needed. The helper receives the incident type, not the location. Parentheses matter: `check_priority(incident_type)` calls the helper; `check_priority` alone refers to the function object.

**Decomposition:** The helper answers one question; the outer procedure decides how to present that answer. Adding a new report requires a new call rather than a copied condition and pair of messages. This simplifies maintenance without claiming an automatic speed improvement.

## College Board pseudocode connection

Python's `def` corresponds to an AP `PROCEDURE` definition. A call supplies arguments in parameter order, and `RETURN(expression)` sends its result back to the calling code. This AP sketch isolates the same helper decision:

```text
PROCEDURE check_priority(incident_type)
{
    RETURN(incident_type = "Medical Emergency")
}
priority ← check_priority("Medical Emergency")
DISPLAY(priority)
priority ← check_priority("Power Outage")
DISPLAY(priority)
```

The results are `TRUE`, then `FALSE`. AP uses `←` for assignment and `=` for equality; Python uses `=` and `==` respectively. In the homework, that returned Boolean is used inside the outer procedure's `if`, rather than being displayed on its own.
