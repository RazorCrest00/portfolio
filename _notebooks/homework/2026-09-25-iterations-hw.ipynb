## Popcorn: SFSRC dispatch scanner

My alert sequence is `[0, 3, 1, 5, 7, 10, 4, 8]`. I chose **3** as the action threshold and **8** as the critical threshold. Below 3 means skip; 3–7 means act and continue scanning; 8 or above means act, then stop.

**Prediction:** Inspect the first six readings. Skip `0` and `1`; count `3`, `5`, `7`, and `10` as four action alerts, totaling `25`. The critical `10` counts before the loop stops. The final `4` and `8` remain uninspected.

````python
# CODE_RUNNER: Iterations Popcorn - SFSRC dispatch scanner
alerts = [0, 3, 1, 5, 7, 10, 4, 8]
safe_cutoff = 3
critical_cutoff = 8
checked_count = 0
safe_count = 0
action_count = 0
total_action_score = 0
critical_found = False

for alert in alerts:
    checked_count += 1  # Count safe and action readings, including the critical one.
    if alert < safe_cutoff:
        safe_count += 1
        print("Alert " + str(alert) + ": safe, skipping")
        continue

    action_count += 1
    total_action_score += alert
    print("Alert " + str(alert) + ": action needed")
    if alert >= critical_cutoff:
        critical_found = True
        print("Critical alert found - stopping scan")
        break

print("Checked alerts:", checked_count)
print("Safe alerts skipped:", safe_count)
print("Action alerts:", action_count)
print("Total action score:", total_action_score)
print("Critical found:", critical_found)
print("Uninspected alerts:", len(alerts) - checked_count)
````

### Why the completed blanks work

The iterable is `alerts`; `alert` receives one value on each pass. `+= 1` counts events, while `+= alert` accumulates their severity scores. All counters start before the loop. `continue` bypasses action counting for safe readings; `break` prevents all remaining readings from being inspected. Neither statement ends the entire program: the final report still runs.

**Sanity check:** `checked_count = safe_count + action_count = 2 + 4 = 6`. Total action score is `3 + 5 + 7 + 10 = 25`. Replacing `continue` with `break` would stop immediately at 0; replacing the final `break` with `continue` would incorrectly process the last two alerts.

## Homework: UESL match-latency checker

**Purpose:** A game-session host needs to identify concerning latency readings and pause the scan at the first critical reading. The sample contains **10 latency measurements in milliseconds**. The action threshold is **40 ms**, and the critical threshold is **90 ms**.

**Prediction:** Read `18, 39, 40, 66, 27, 89, 90`, then stop. The final `45, 12, 99` are uninspected, not automatically safe. Expect **7 checked, 3 skipped, 4 actions, 285 ms total action latency, critical found True, and 3 uninspected**. Summing readings is the required accumulator exercise; it is not the latency of one match.

````python
# CODE_RUNNER: Iterations Homework - UESL match-latency checker
match_latencies_ms = [18, 39, 40, 66, 27, 89, 90, 45, 12, 99]
safe_cutoff = 40
critical_cutoff = 90
checked_count = 0
skipped_count = 0
action_count = 0
total_action_latency_ms = 0
critical_found = False

for latency_ms in match_latencies_ms:
    checked_count += 1
    if latency_ms < safe_cutoff:
        skipped_count += 1
        print(str(latency_ms) + " ms: safe, no action")
        continue

    action_count += 1
    total_action_latency_ms += latency_ms
    print(str(latency_ms) + " ms: UESL host action needed")
    if latency_ms >= critical_cutoff:
        critical_found = True
        print("Critical match latency found. Stop checking.")
        break

print("UESL Match Latency - Final Report")
print("Checked readings:", checked_count)
print("Skipped safe readings:", skipped_count)
print("Action readings:", action_count)
print("Total action latency (ms):", total_action_latency_ms)
print("Critical found:", critical_found)
print("Uninspected readings:", len(match_latencies_ms) - checked_count)
````

| 39 | Skip | 2 | 2 | 0 | 0 | False |
| 40 | Action: equals lower threshold | 3 | 2 | 1 | 40 | False |
| 66 | Action | 4 | 2 | 2 | 106 | False |
| 27 | Skip | 5 | 3 | 2 | 106 | False |
| 89 | Action: below critical threshold | 6 | 3 | 3 | 195 | False |
| 90 | Action, then stop: equals critical threshold | 7 | 3 | 4 | 285 | True |

**Order matters:** Count the visit before `continue`, and count/add the critical reading before `break`. Initializing a counter inside the loop would erase earlier work. Using `<= 40` would incorrectly skip 40; using `> 90` would miss the critical boundary at 90.

**Invariants:** Every inspected reading is classified exactly once, so checked = skipped + actions. Checked + uninspected = list length. The action total contains only inspected action readings, including the first critical one. The input list is read without modification.

**Loop choice and termination:** A `for` loop fits the finite input list. It ends after all values or at the first critical reading. Early exit intentionally leaves a suffix unread; its report does not characterize the entire dataset. The scan takes at most `n` iterations and uses a fixed number of counters beyond the input list. It reduces repetitive code without assuming every loop-based implementation is faster.

## Extension: Condition-controlled repetition

The homework requires `for`, `continue`, and `break`; this additional example demonstrates `while`. A simulated game-render queue starts at 31 frames. Each processing tick removes 7 until **at most 10** remain. Predict `24 → 17 → 10`, three ticks, and a final false condition. Using `!= 10` instead of `> 10` could fail to terminate when a step skips over 10.

````python
# CODE_RUNNER: Iterations Extension - Drain a game-render queue
queued_frames = 31
target_queue = 10
frames_per_tick = 7
ticks = 0

if queued_frames < 0 or target_queue < 0:
    raise ValueError("frame counts and the target must be nonnegative")

if frames_per_tick <= 0:
    raise ValueError("frames_per_tick must be positive so the queue shrinks")

while queued_frames > target_queue:
    queued_frames = max(0, queued_frames - frames_per_tick)
    ticks += 1
    print("Queued frames:", queued_frames)

print("Processing ticks:", ticks)
print("Final queued frames:", queued_frames)
print("Target reached:", queued_frames <= target_queue)
````

### Fixed-count repetition and off-by-one checks

A short rehearsal needs exactly three readiness checks. `range(1, checks_needed + 1)` includes 1 through 3 because Python excludes the stop value. The matching AP structure is `REPEAT 3 TIMES`. With `checks_needed = 0`, no check runs and the final count remains zero.

````python
# CODE_RUNNER: Iterations Extension - Three UESL readiness checks
checks_needed = 3
completed_checks = 0

for check_number in range(1, checks_needed + 1):
    print("UESL readiness check:", check_number)
    completed_checks += 1

print("Completed checks:", completed_checks)
````

## College Board pseudocode connection

This is the homework's **equivalent scan logic** in AP notation. AP lists start at 1. `REPEAT UNTIL` stops when its condition becomes true; Python `while` continues while its condition is true. The official reference provides neither Python's `break` nor `continue`, so selection handles safe readings and a Boolean flag handles early exit. [College Board reference sheet, iteration and list operations](https://apcentral.collegeboard.org/media/pdf/ap-computer-science-principles-exam-reference-sheet.pdf#page=4)

```text
latencies ← [18, 39, 40, 66, 27, 89, 90, 45, 12, 99]
safeCutoff ← 40
criticalCutoff ← 90
i ← 1
checked ← 0
skipped ← 0
actions ← 0
total ← 0
criticalFound ← FALSE
REPEAT UNTIL(i > LENGTH(latencies) OR criticalFound)
{
    latency ← latencies[i]
    checked ← checked + 1
    IF(latency < safeCutoff)
    {
        skipped ← skipped + 1
    }
    ELSE
    {
        actions ← actions + 1
        total ← total + latency
        IF(latency ≥ criticalCutoff)
        {
            criticalFound ← TRUE
        }
    }
    i ← i + 1
}
DISPLAY(checked)
DISPLAY(skipped)
DISPLAY(actions)
DISPLAY(total)
DISPLAY(criticalFound)
DISPLAY(LENGTH(latencies) - checked)
```

The six report values are `7, 3, 4, 285, TRUE, 3`, matching the Python report's values. With an empty list, the stop condition is already true and no index is read. Incrementing `i` ensures progress even for a safe reading. The Python `for` loop handles that progress automatically.
