## Homework — Readings and calculator layout

The assignment asks for the calculator layout, not calculator arithmetic. The title is a full-row header; Beaker 3 and Avg are the two accents. All eight boxes have the base cell class.

{% capture sass_saved_output %}

```python
%%html
<!-- UI_RUNNER: OCS Grids — Homework — Readings and calculator layout -->
<div id="sass-grids-exercise-3">
<div class="ocs__grid ocs__grid--standard cols-3">
  <div class="ocs__grid-cell ocs__grid-cell--header">Temperature Trial (°C)</div>
  <div class="ocs__grid-cell">Beaker 1: 22.4</div>
  <div class="ocs__grid-cell">Beaker 2: 22.9</div>
  <div class="ocs__grid-cell ocs__grid-cell--accent">Beaker 3: 31.6</div>
</div>
<div class="ocs__grid ocs__grid--calculator">
  <div class="ocs__grid-cell">7</div>
  <div class="ocs__grid-cell">8</div>
  <div class="ocs__grid-cell">9</div>
  <div class="ocs__grid-cell ocs__grid-cell--accent">Avg</div>
</div>
<script>
const root = document.getElementById('sass-grids-exercise-3');
const grids = [...root.querySelectorAll('.ocs__grid')];
const cells = [...root.querySelectorAll('.ocs__grid-cell')];
const classes = [...root.querySelectorAll('[class]')].flatMap(e => [...e.classList]);
const checks = [
  ['No inline style attributes', !root.querySelector('[style]')],
  ['No made-up classes', classes.every(c => c.startsWith('ocs__') || c.startsWith('cols-'))],
  ['Two grids with exactly one variant each', grids.length === 2 && grids.every(g => [...g.classList].filter(c => c.startsWith('ocs__grid--')).length === 1)],
  ['Standard variant for Grid 1', grids[0]?.classList.contains('ocs__grid--standard')],
  ['Calculator variant for Grid 2', grids[1]?.classList.contains('ocs__grid--calculator')],
  ['Every box is a grid cell', cells.length === 8 && grids.every(g => [...g.children].every(c => c.classList.contains('ocs__grid-cell')))],
  ['Title is a header cell', grids[0]?.firstElementChild.classList.contains('ocs__grid-cell--header')],
  ['Outlier and Avg are accents', grids[0]?.lastElementChild.classList.contains('ocs__grid-cell--accent') && grids[1]?.lastElementChild.classList.contains('ocs__grid-cell--accent')]
];
const report = document.createElement('ul');
report.setAttribute('aria-label', 'Homework checks');
for (const [label, pass] of checks) {
  const item = document.createElement('li');
  item.textContent = (pass ? 'PASS — ' : 'FIX — ') + label;
  report.append(item);
}
root.prepend(report);
</script>
</div>
```
