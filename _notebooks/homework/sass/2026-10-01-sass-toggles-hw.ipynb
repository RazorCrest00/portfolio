## Homework — UESL rehearsal controls

Each label names an independent feature and wraps a native checkbox. aria-controls points to the matching panel. One rendering function derives both visibility and the live count from actual state. Reset returns every feature to off.

{% capture sass_saved_output %}

```python
%%html
<!-- UI_RUNNER: OCS Toggles — Homework — UESL rehearsal controls -->
<div id="sass-toggles-exercise-2">
<section id="uesl-settings" aria-labelledby="uesl-title">
  <h2 id="uesl-title">UESL rehearsal features</h2>
  <p>Choose which aids to display. This demo keeps no personal data.</p>
  <p><label class="ocs__toggle ocs__toggle--green">
    <input id="uesl-captions" type="checkbox" class="ocs__toggle-input" aria-controls="uesl-caption-panel">
    <span class="ocs__toggle-track" aria-hidden="true"></span>
    <span class="ocs__toggle-label">Show match captions</span>
  </label></p>
  <p><label class="ocs__toggle ocs__toggle--green">
    <input id="uesl-controls" type="checkbox" class="ocs__toggle-input" aria-controls="uesl-controls-panel">
    <span class="ocs__toggle-track" aria-hidden="true"></span>
    <span class="ocs__toggle-label">Show controls guide</span>
  </label></p>
  <p><label class="ocs__toggle ocs__toggle--green">
    <input id="uesl-debug" type="checkbox" class="ocs__toggle-input" aria-controls="uesl-debug-panel">
    <span class="ocs__toggle-track" aria-hidden="true"></span>
    <span class="ocs__toggle-label">Show developer readout</span>
  </label></p>
  <p id="uesl-count" role="status" aria-live="polite">0 of 3 features enabled</p>
  <section id="uesl-caption-panel" hidden><h3>Match captions</h3><p>Blue team captures the flag — sample caption.</p></section>
  <section id="uesl-controls-panel" hidden><h3>Controls guide</h3><p>WASD: move. Space: jump. Escape: pause.</p></section>
  <section id="uesl-debug-panel" hidden><h3>Developer readout</h3><p>Scene: practice arena. Build: classroom prototype.</p></section>
  <button id="uesl-reset" type="button" class="ocs__btn outline">Reset features</button>
</section>
<script>
const root = document.getElementById('uesl-settings');
const toggles = [...root.querySelectorAll('input[type="checkbox"]')];
const count = document.getElementById('uesl-count');
function renderFeatures() {
  // .checked is the live Boolean state; the checked HTML attribute is only a default.
  for (const toggle of toggles) {
    const panel = document.getElementById(toggle.getAttribute('aria-controls'));
    // hidden removes an inactive feature from layout and the accessibility tree.
    panel.hidden = !toggle.checked;
  }
  // Recompute from the controls rather than incrementing a counter that can drift.
  const enabled = toggles.filter(toggle => toggle.checked).length;
  count.textContent = enabled + ' of ' + toggles.length + ' features enabled';
}
toggles.forEach(toggle => toggle.addEventListener('change', renderFeatures));
document.getElementById('uesl-reset').addEventListener('click', () => {
  // Programmatic changes do not emit change, so reset must render explicitly.
  toggles.forEach(toggle => { toggle.checked = false; });
  renderFeatures();
});
// Initialize once so restored/default control state and visible features agree.
renderFeatures();
</script>
</div>
```
