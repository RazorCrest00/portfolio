## Popcorn 2: Fix the unreachable branches

**Original prediction:** Score 9 incorrectly prints “Be a little careful, but this is probably fine.” The `>= 6` and `>= 9` branches can never run because every value meeting either already meets `>= 3`. Test highest thresholds first. Boundary scores 3, 6, and 9 belong to the new band, not the previous one.


````python
# CODE_RUNNER: 3.06 Popcorn 2 - Reachable risk bands
def report_risk(risk_score):
    # 3 thresholds -> 4 outcomes -> 4 answers
    if risk_score >= 9:
        print("Do not reply. This is almost certainly a scam.")
    elif risk_score >= 6:
        print("This looks risky. Check with someone you trust before replying.")
    elif risk_score >= 3:
        print("Be a little careful, but this is probably fine.")
    else:
        print("Nothing suspicious in this message.")

for risk_score in [0, 3, 6, 9]:
    print("Risk score:", risk_score)
    report_risk(risk_score)
````

## Homework: Score first, then advise

**Prediction:** The supplied gift-card message triggers all three signals: `4 + 3 + 2 = 9`, reaching the highest band. Independent `if` statements allow signals to accumulate; an `elif` between signals would incorrectly stop after the first match. Every completed evaluation then reaches exactly one reporting answer. This is a classroom scoring rule, not a guarantee that a message is safe or fraudulent.


````python
# CODE_RUNNER: 3.06 Homework - Message scoring and advice
def score_message(message, amount_requested, known_contact):
    risk_score = 0
    # No else is needed: an absent signal contributes zero to the existing score.
    if "gift cards" in message.lower():
        risk_score = risk_score + 4
    # No else is needed: amounts at or below 100 leave the score unchanged.
    if amount_requested > 100:
        risk_score = risk_score + 3
    # No else is needed: a known sender contributes zero, not a negative score.
    if not known_contact:
        risk_score = risk_score + 2
    return risk_score

def report_risk(risk_score):
    # 3 thresholds -> 4 outcomes -> 4 answers
    if risk_score >= 9:
        print("Do not reply. This is almost certainly a scam.")
    elif risk_score >= 6:
        print("This looks risky. Check with someone you trust before replying.")
    elif risk_score >= 3:
        print("Be a little careful, but this is probably fine.")
    else:
        print("Nothing suspicious in this message.")

message = "URGENT: send gift cards now to unlock your account"
amount_requested = 500
known_contact = False
risk_score = score_message(message, amount_requested, known_contact)
print("Risk score:", risk_score)
report_risk(risk_score)
````

## Reachability proof and extreme tests

| Inputs: message / amount / known contact | Score | Reachable answer |
| --- | ---: | --- |
| Meet at practice / 0 / True | 0 | Nothing suspicious |
| Equipment payment / 101 / True | 3 | Be a little careful |
| Please send gift cards / 0 / False | 6 | Check with someone you trust |
| URGENT: send gift cards / 500 / False | 9 | Do not reply |

These cover all four reporting paths. The first and last are the clean and all-signals extremes. The extra boundary cases show that 100 does not add points, 101 does, and changing capitalization does not hide the phrase. The reporting chain covers all scores from zero upward, including scores not produced by these three flags.

**Stretch explanation:** A scoring `if` may do nothing because zero contribution is intentional; a reporting `if` without a fallback leaves the user without an answer.


````python
# CODE_RUNNER: 3.06 Tests - Four branches and boundaries
def score_message(message, amount_requested, known_contact):
    risk_score = 0
    # No else is needed: an absent signal contributes zero to the existing score.
    if "gift cards" in message.lower():
        risk_score = risk_score + 4
    # No else is needed: amounts at or below 100 leave the score unchanged.
    if amount_requested > 100:
        risk_score = risk_score + 3
    # No else is needed: a known sender contributes zero, not a negative score.
    if not known_contact:
        risk_score = risk_score + 2
    return risk_score

def report_risk(risk_score):
    # 3 thresholds -> 4 outcomes -> 4 answers
    if risk_score >= 9:
        print("Do not reply. This is almost certainly a scam.")
    elif risk_score >= 6:
        print("This looks risky. Check with someone you trust before replying.")
    elif risk_score >= 3:
        print("Be a little careful, but this is probably fine.")
    else:
        print("Nothing suspicious in this message.")

cases = [
    ("clean", "Meet at practice", 0, True, 0),
    ("caution", "Equipment payment", 101, True, 3),
    ("risky", "Please send gift cards", 0, False, 6),
    ("all signals", "URGENT: send gift cards", 500, False, 9),
    ("exact amount boundary", "Meet at practice", 100, True, 0),
    ("capitalized phrase", "SEND GIFT CARDS", 0, True, 4),
]
for name, message, amount, known, expected in cases:
    score = score_message(message, amount, known)
    assert score == expected, name
    print(name + ":", score)
    report_risk(score)
````
